```python
import sqlite3

conn = sqlite3.connect("chargepoint_md.db")
cur = conn.cursor()

# Charging station tiles (dynamic)
cur.execute("""
CREATE TABLE IF NOT EXISTS ChargingSession (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    timestamp TEXT,
    lat REAL,
    lon REAL,
    available INTEGER,
    in_use INTEGER,
    other INTEGER
)
""")

conn.commit()
conn.close()
```
